```python
# Load all the necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

import SimpleITK as sitk

from radiomics import imageoperations as io
```

```python
nuc_df = pd.read_csv(
    "../results/pyradiomics/nobin_0.1_z_score_all_df.csv",
    index_col=0,
)

def from_zxy_to_xyz(array_list):
    """Transpose the image dimensions from ZXY to XYZ coordinates"""
    return [np.transpose(a, axes=[1, 2, 0]) for a in array_list]
```

```python
my_df = nuc_df
condition = my_df.condition

czi_metadata_filt = pd.read_csv("../results/czi_metadata_filt.csv")
my_df = my_df.merge(czi_metadata_filt, left_on="czi_path", right_on="path")
my_df["condition"] = condition
my_df = my_df[my_df["condition"].isin(["young", "aged"])]
my_df = my_df.reset_index(drop=True)
```

```python
aged_df = my_df[my_df["condition"] == "aged"]
young_df = my_df[my_df["condition"] == "young"]
```

```python
filt = young_df["log-sigma-0-5-mm-3D_glcm_SumEntropy"] < 1.538
folder = "../../data"

young_df[filt]["npz_path"].shape
```

```python
for npz in young_df[filt]["npz_path"]:

    print(npz)
    image = np.load(f"{folder}/{npz}", allow_pickle=True)
    im = image["img"]
    nuc_mask = image["nuc_mask"].astype(int)

    im, nuc_mask = from_zxy_to_xyz([im, nuc_mask])
    im = sitk.GetImageFromArray(im)
    nuc_mask = sitk.GetImageFromArray(nuc_mask)
    filt_im = io.getLoGImage(im, nuc_mask, sigma=[0.5])

    plt.figure(figsize=(10,4))
    fig, axes = plt.subplots(3, 10, figsize=(10, 4))
    ax_idx = 0
    
    for this_im, filt_name, _ in filt_im:
        print(filt_name)
        if filt_name == "log-sigma-0-5-mm-3D":
            
            this_im = sitk.GetArrayFromImage(this_im)
            nuc_mask = sitk.GetArrayFromImage(nuc_mask)
            pic_range = np.linspace(0, this_im.shape[2] - 1, 30, dtype=int)
            
            for i, pic in enumerate(pic_range):

                row, col = divmod(ax_idx, 10)
                plot_im = this_im[:, :, pic]*nuc_mask[:, :, pic]
                plt.imshow(plot_im, cmap="RdBu")
                axes[row, col].imshow(plot_im, cmap="RdBu")
                axes[row, col].set_title(str(pic))
                axes[row, col].axis("off")
                ax_idx += 1
        
    plt.show()
```

```python
for npz in young_df[filt]["npz_path"]:

    print(npz)
    image = np.load(f"{folder}/{npz}", allow_pickle=True)
    im = image["img"]
    nuc_mask = image["nuc_mask"].astype(int)

    im, nuc_mask = from_zxy_to_xyz([im, nuc_mask])
    max_index = np.argmax(nuc_mask.sum(0).sum(0))
    plt.imshow(im[:, :, max_index])
    plt.title(str(max_index))
    plt.axis("off")
    plt.show()
    
    im = sitk.GetImageFromArray(im)
    nuc_mask = sitk.GetImageFromArray(nuc_mask)
    filt_im = io.getLoGImage(im, nuc_mask, sigma=[0.5])

    plt.figure(figsize=(10,4))
    
    for this_im, filt_name, _ in filt_im:
        if filt_name == "log-sigma-0-5-mm-3D":
            
            this_im = sitk.GetArrayFromImage(this_im)
            nuc_mask = sitk.GetArrayFromImage(nuc_mask)
            plot_im = this_im[:, :, max_index]*nuc_mask[:, :, max_index]
            plt.imshow(plot_im, cmap="RdBu")
            plt.title(f"Z={str(max_index)}")
            plt.colorbar()
            plt.axis("off")
        
    plt.show()
```
